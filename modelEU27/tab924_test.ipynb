```python
import random
def making924table(seed, rankNum):
    tab = [[0] * rankNum for _ in range(924)]
    random.seed(seed)
    for i in range(924):
        if i==2 and rankNum > 2:
            r1=0
            r2=0
            r3=0
            while r1==r2 or r1==r3 or r2==r3:
                r1=random.randint(0,rankNum-1)
                r2=random.randint(0,rankNum-1)
                r3=random.randint(0,rankNum-1)
            tab[i][r1] = 1/3
            tab[i][r2] = 1/3
            tab[i][r3] = 1/3
        else:
            tab[i][random.randint(0,rankNum-1)] = 1
        
    return tab   
```

```python
numRank=40
tab924 = making924table(12345,numRank)
```

```python
for i in range(924):
    for j in range(numRank):
        if tab924[i][j]==1/3: print(i,j)
```

```python
#count positions in tab
countInRanks = [0] * numRank
for i in range(924):
    for j in range(numRank):
        if tab924[i][j] > 0: countInRanks[j]+=1
countInRanks
```

```python
import statistics as s
s.mean(countInRanks)
```

```python
s.stdev(countInRanks)
```
